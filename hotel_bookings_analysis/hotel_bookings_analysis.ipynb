```python
# Приведем названия столбцов к нижнему регистру
# Заменим пробелы в названиях на нижнее подчеркивание

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
# Узнаем пользователи из каких стран совершили наибольшее число успешных бронирований.
# Выведем ТОП-5

top_countries = (df.query('is_canceled == 0')
.groupby('country', as_index=False)
.agg(count_booking=('is_canceled', 'count'))
.sort_values('count_booking', ascending=False)
.head())
top_countries
```

```python
# Узнаем на сколько ночей в среднем бронируют отели разных типов

mean_nights_at_hotel = (df.groupby('hotel', as_index=False)
.agg(mean_nights=('stays_total_nights', 'mean')))
mean_nights_at_hotel
```

```python
# Узнаем сколько раз тип номера, полученный клиентом, отличался от изначально забронированного

df.query('assigned_room_type != reserved_room_type').shape[0]
```

```python
# Узнаем на какой месяц чаще всего оформляли бронь в 2016 году

most_booking_month_2016 = (
    df.query('arrival_date_year == 2016')
    .groupby('arrival_date_month', as_index=False)
    .agg(count_booking=('hotel', 'count'))
    .sort_values('count_booking', ascending=False).head(1)
)
most_booking_month_2016
```

```python
# Узнаем на какой месяц чаще всего оформляли бронь в 2017 году

most_booking_month_2017 = (
    df.query('arrival_date_year == 2017')
    .groupby('arrival_date_month', as_index=False)
    .agg(count_booking=('hotel', 'count'))
    .sort_values('count_booking', ascending=False).head(1)
)
most_booking_month_2017
```

```python
# Узнаем на какой месяц бронирования отеля типа City Hotel отменялись чаще всего в 2015 году

df_canceled = (
    df.query("hotel == 'City Hotel' and is_canceled == 1")
    .groupby(['arrival_date_year', 'arrival_date_month'], as_index=False)['hotel'].count()
)
df_canceled_2015 = df_canceled.query('arrival_date_year == 2015').sort_values('hotel', ascending=False).head(1)
df_canceled_2015
```

```python
# Узнаем на какой месяц бронирования отеля типа City Hotel отменялись чаще всего в 2016 году

df_canceled_2016 = df_canceled.query('arrival_date_year == 2016').sort_values('hotel', ascending=False).head(1)
df_canceled_2016
```

```python
# Узнаем на какой месяц бронирования отеля типа City Hotel отменялись чаще всего в 2017 году

df_canceled_2017 = df_canceled.query('arrival_date_year == 2017').sort_values('hotel', ascending=False).head(1)
df_canceled_2017
```

```python
# Найдем наибольшее среднее значение среди 'adults', 'children', 'babies'

df[['adults', 'children', 'babies']].mean().idxmax()
```

```python
# Создадим колнку total_kids
# Узнаем для отелей какого типа среднее значение total_kids оказалось наибольшим

df['total_kids'] = df['children'] + df['babies']
df.groupby('hotel', as_index=False)['total_kids'].mean().round(2).sort_values('total_kids', ascending=False)
```

```python
# Создадим колонку has_kids
# Посчитаем Churn Rate - процент клиентов которые отменили бронирование. Проверим как связана эта метрика с наличием детей

df['has_kids'] = df['total_kids'] > 0
df.groupby('has_kids')['is_canceled'].mean().round(4) * 100
```
